## IMPORTING DATASET

```python
import pandas as pd
df=pd.read_csv("C:\\Users\goura\Downloads\Student_Performance.csv")
```

```python
df.head()
```

## PREPROCESSING THE DATA
the task asks us to predict whether a student will pass or fail based on the number of hours he/she studies,attendence and marks.
so this dataset needs to be minimised and preprocessed.

```python
# selecting a handful of columns columns 
df = df[['student_id', 'study_hours', 'attendance_percentage', 'overall_score', 'final_grade']]
```

```python
df.head()
```

we are trying to predict pass or fail but the final grade columns has ranges varying from a-f, so
i think its better to leave F as 0 and  then try to classify the rest as 1 (a-e) 

```python
df['final_grade'] = df['final_grade'].replace('f', 0)
df['final_grade'] = df['final_grade'].replace(['a','b','c','d','e'], 1)
```

```python
df.head()
```

```python
print(df)
```

```python
# taking data as input and output
inp=df.iloc[:,:4]
out=df.iloc[:,4:]
```

```python
out.head()
```

## train-test split
splitting the data for training and testing and scaling them

```python
from sklearn.model_selection import train_test_split as tts
inp_tr,inp_te,out_tr,out_te=tts(inp,out,test_size=0.3)
```

```python
out_tr
```

## SCALING

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
inp_tr=ss.fit_transform(inp_tr)
```

```python
inp_tr
```

```python
inp_te=ss.fit_transform(inp_te)
```

## MODEL TRAINING
using KNN and logistic regression for this dataset

```python
from sklearn.linear_model import LogisticRegression
BP=LogisticRegression()
BP.fit(inp_tr,out_tr)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(inp_tr,out_tr)
```

## TESTING

```python
BP.predict(inp_te)
```

```python
knn.predict(inp_te)
```

```python
from sklearn.metrics import accuracy_score as acs
BPT=BP.predict(inp_te)
KNNT=knn.predict(inp_te)
acs1=acs(BPT,out_te)
acs2=acs(KNNT,out_te)
print(BPT)
print(KNNT)
print(acs1)
print(acs2)
```

```python
print(BP.coef_)
```

## VISUALIZATION

```python
from sklearn.metrics import confusion_matrix
CM1=confusion_matrix(out_te,BPT)
CM2=confusion_matrix(out_te,KNNT)
print(CM1)
print(CM2)
```
